```python

# Conversation Turn-Taking Analyzer

# Sample conversation segments
conversation = [
    {"speaker": "Speaker A", "start": 0, "end": 4},
    {"speaker": "Speaker B", "start": 3.5, "end": 6},
    {"speaker": "Speaker A", "start": 6.5, "end": 8},
    {"speaker": "Speaker C", "start": 8, "end": 11},
    {"speaker": "Speaker B", "start": 10.5, "end": 13},
    {"speaker": "Speaker C", "start": 13.5, "end": 15}
]

speaking_time = {}
turn_count = {}
interruptions = 0

# Calculate speaking time and turns
for segment in conversation:
    speaker = segment["speaker"]
    duration = segment["end"] - segment["start"]

    speaking_time[speaker] = (
        speaking_time.get(speaker, 0) + duration
    )
    turn_count[speaker] = turn_count.get(speaker, 0) + 1

# Detect overlapping speech between consecutive segments
for i in range(1, len(conversation)):
    previous = conversation[i - 1]
    current = conversation[i]

    if (current["speaker"] != previous["speaker"]
            and current["start"] < previous["end"]):
        interruptions += 1

total_speaking_time = sum(speaking_time.values())

print("=" * 45)
print("CONVERSATION TURN-TAKING ANALYZER")
print("=" * 45)

for speaker in sorted(speaking_time):
    percentage = (
        speaking_time[speaker] / total_speaking_time
    ) * 100

    print(f"\n{speaker}")
    print(f"Speaking time: {speaking_time[speaker]:.1f} seconds")
    print(f"Turns: {turn_count[speaker]}")
    print(f"Speaking share: {percentage:.1f}%")

print("\nOverlapping speech instances:", interruptions)
print("Total recorded speaking time:",
      f"{total_speaking_time:.1f} seconds")
print("=" * 45)
```
